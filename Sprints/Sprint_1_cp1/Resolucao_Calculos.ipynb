Calcular o deslocamento de água, a posição do centro de gravidade (CG), centro de carena (CB) e altura metacêntrica inicial (GM) para massa total de 85 kg.

### Modelagem Matemática Formal
$$\text{Empuxo: } E = \rho_{\text{água}} \cdot g \cdot V_{\text{submerso}}, \quad \overline{GM} = \overline{KB} + \overline{BM} - \overline{KG} > 0$$


### Passo 1: Volume Submerso Requerido para Flutuação
Para massa total M = 85 kg em água doce (\rho = 1.000\text{ kg/m}^3): Volume submerso necessário: V_{sub} = \frac{85}{1.000} = 0,085\text{ m}^3 (85 litros no total, 42,5 litros por pontão).


### Passo 2: Geometria dos Dois Pontões do Catamarã
Comprimento L = 1,60 m; largura de cada pontão b = 0,22 m; calado de projeto T = 0,135 m. Volume total dos pontões = 2 \times (1,60 \times 0,22 \times 0,25) = 0,176\text{ m}^3, fornecendo reserva de flutuabilidade de 107%.


### Passo 3: Momento de Inércia da Linha de Água do Catamarã
A separação entre eixos dos pontões é d = 0,70 m: I_{transversal} \approx 2 \times \left[ \frac{L \cdot b^3}{12} + L \cdot b \cdot (d/2)^2 \right] \approx 0,0875\text{ m}^4.


### Passo 4: Cálculo do Raio Metacêntrico (BM) e Estabilidade GM
\overline{BM} = \frac{I}{V_{sub}} = \frac{0,0875}{0,085} \approx 1,029\text{ m}. Com \overline{KB} = T/2 = 0,0675\text{ m} e \overline{KG} = 0,35\text{ m}: \overline{GM} = 0,0675 + 1,029 - 0,35 = 0,7465\text{ m}.


### Passo 5: Conclusão de Navegabilidade e Segurança Náutica
A altura metacêntrica de 0,75 m confere excepcional rigidez contra emborcamento mesmo sob ondas e correntezas fluviais de até 1,5 m/s.


```python
import numpy as np
import matplotlib.pyplot as plt

angulo_banda = np.linspace(0, 30, 100) # Angulo de inclinacao em graus
ang_rad = np.radians(angulo_banda)
gm = 0.7465 # m
m = 85.0
g = 9.81
braco_endireitante_gz = gm * np.sin(ang_rad)
torque_restaurador = m * g * braco_endireitante_gz
fig, ax = plt.subplots(figsize=(6, 3.5), facecolor='#12193b')
ax.set_facecolor('#0a0e24')
ax.plot(angulo_banda, torque_restaurador, color='#06b6d4', lw=2.5, label='Torque Restaurador de Estabilidade (N·m)')
ax.set_title('Curva de Estabilidade Hidrodinâmica do Catamarã USV', color='#ffffff', fontsize=10.5, fontweight='bold')
ax.set_xlabel('Ângulo de Adernamento / Banda (Graus)', color='#94a3b8', fontsize=9)
ax.set_ylabel('Momento Restaurador (N·m)', color='#94a3b8', fontsize=9)
ax.tick_params(colors='#94a3b8')
ax.grid(True, color='#2d3e87', alpha=0.5, ls=':')
ax.legend(facecolor='#1c2759', edgecolor='#2d3e87', labelcolor='#ffffff', fontsize=8)
plt.tight_layout()
plt.savefig('grafico_resolucao.png', dpi=200, facecolor=fig.get_facecolor())
plt.close()
```

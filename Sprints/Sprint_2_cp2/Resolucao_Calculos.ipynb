> **Especificação da Sprint:**
> Dimensionar o sistema de propulsão elétrica baseado em dois propulsores subaquáticos (Thrusters BLDC) com empuxo nominal combinado de 14 kgf para vencer o arrasto hidrodinâmico à velocidade de cruzeiro de v = 1,8 m/s (3,5 nós). Dimensionar o teto de painéis solares semiflexíveis monocristalinos de 150 W instalados sobre a embarcação e o banco de baterias LiFePO4 de 24 V / 40 Ah.

### Modelagem Matemática Formal
$$R_{\text{arrasto}} = \frac{1}{2} C_d \cdot \rho_{\text{água}} \cdot S_{\text{molhada}} \cdot v^2, \quad P_{\text{elétrica}} = \frac{R_{\text{arrasto}} \cdot v}{\eta_{\text{prop}}}$$


### Passo 1: Cálculo da Resistência Hidrodinâmica ao Avanço
Para área molhada S = 0,92 m², coeficiente de arrasto de casco esguio C_d = 0,06 e v = 1,8 m/s: R = \frac{1}{2} \times 0,06 \times 1.000 \times 0,92 \times (1,8)^2 = 30 \times 0,92 \times 3,24 \approx 89,42\text{ N} (\approx 9,12\text{ kgf}).


### Passo 2: Potência Propulsiva Útil e Potência Elétrica
Potência mecânica: P_{mec} = 89,42 \times 1,8 \approx 160,96\text{ W}. Com rendimento do propulsor marinho \eta = 55%: P_{eletr} = \frac{160,96}{0,55} \approx 292,65\text{ W} (146,3 W por motor).


### Passo 3: Contribuição da Geração Solar de Topo
Painel solar de 150 W entregando média de 115 W úteis durante período diurno (10h às 16h), reduzindo o consumo líquido extraído da bateria para 292,65 - 115 = 177,65 W.


### Passo 4: Autonomia Contínua em Missão Diurna
Com banco de baterias de 24 V \times 40 Ah = 960 Wh (DoD 85% = 816 Wh): Autonomia = \frac{816}{177,65} \approx 4,6\text{ horas} a velocidade plena de 3,5 nós.


### Passo 5: Conclusão Operacional
Capacidade de varredura autônoma de até 30 km de curso d'água por missão diária completa.


```python
import numpy as np
import matplotlib.pyplot as plt

vel_nos = np.linspace(1.0, 5.0, 100)
vel_ms = vel_nos * 0.5144
r_arrasto = 0.5 * 0.06 * 1000 * 0.92 * (vel_ms**2)
p_eletrica = (r_arrasto * vel_ms) / 0.55
fig, ax = plt.subplots(figsize=(6, 3.5), facecolor='#12193b')
ax.set_facecolor('#0a0e24')
ax.plot(vel_nos, p_eletrica, color='#818cf8', lw=2.5, label='Potência Propulsiva Requerida (W)')
ax.axvline(3.5, color='#f59e0b', ls='--', label='Velocidade de Cruzeiro (3.5 nós -> 293 W)')
ax.axhline(150, color='#10b981', ls=':', lw=1.5, label='Geração Painel Solar (150 W)')
ax.set_title('Curva de Potência Elétrica de Propulsão vs. Velocidade Náutica', color='#ffffff', fontsize=10.5, fontweight='bold')
ax.set_xlabel('Velocidade de Cruzeiro da Embarcação (Nós)', color='#94a3b8', fontsize=9)
ax.set_ylabel('Potência Elétrica Consumida (Watts)', color='#94a3b8', fontsize=9)
ax.tick_params(colors='#94a3b8')
ax.grid(True, color='#2d3e87', alpha=0.5, ls=':')
ax.legend(facecolor='#1c2759', edgecolor='#2d3e87', labelcolor='#ffffff', fontsize=8)
plt.tight_layout()
plt.savefig('grafico_resolucao.png', dpi=200, facecolor=fig.get_facecolor())
plt.close()
```
